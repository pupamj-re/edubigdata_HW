# 高雄市國小資料分析

本 notebook 示範如何透過檔案路徑匯入並呼叫 `gemini-code-1786348741522.py` 中的函式（因檔名含 `-`，不能直接用 `import`）。
執行後會在 `output/` 產生圖檔。

```python
import importlib.util
import pathlib

# 載入含有破折號的 script 檔案為模組 (module name 使用臨時名稱 'script_mod')
script_path = pathlib.Path('gemini-code-1786348741522.py').resolve()
spec = importlib.util.spec_from_file_location('script_mod', str(script_path))
script_mod = importlib.util.module_from_spec(spec)
spec.loader.exec_module(script_mod)

# 現在可以使用 script_mod.load_data, compute_metrics, make_plots, main 等函式
print('module loaded:', script_mod)
```

```python
# 範例：讀取資料、計算指標、產生圖檔（會輸出到 output/）
df = script_mod.load_data()
df = script_mod.compute_metrics(df)
script_mod.make_plots(df)
print('done')
```

也可以直接呼叫 `script_mod.main()`，或是在本機/CI 中執行 `python gemini-code-1786348741522.py`。
